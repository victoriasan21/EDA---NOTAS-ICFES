```python
# Celda 1: Importar librerías y montar Drive
import pandas as pd
import numpy as np
from google.colab import drive

# Montar Google Drive
drive.mount('/content/drive', force_remount=True)
```

```python
# Celda 2: Definir la ruta del archivo Excel
ruta_archivo_xlsx = '/content/drive/Shareddrives/GESTIÓN DE PROYECTOS PARA ANALÍTICA/Resultados_únicos_Saber_11_20260828 Original.xlsx'
```

```python
# Celda 3: Cargar el dataset desde Excel
df = pd.read_excel(ruta_archivo_xlsx)

print("Dataset cargado exitosamente.")
print("Número de columnas:", len(df.columns))
print("Primeras columnas:", df.columns[:10])
df.head()
```

```python
# Celda 4: Eliminar columnas vacías y las que se llaman "Unnamed"
df = df.dropna(axis=1, how='all')  # borra columnas 100% vacías
df = df.loc[:, df.columns.str.strip() != '']  # borra columnas con nombre vacío
df = df.loc[:, ~df.columns.str.contains('^Unnamed')]  # borra columnas "Unnamed"

print("Número de columnas después de limpieza:", len(df.columns))
print("Primeras columnas:", df.columns[:10])
df.head()
```

```python
# Celda 5: Ajustes de datos

# Convertir códigos DANE a texto (evitar notación científica)
df['COLE_COD_DANE_ESTABLECIMIENTO'] = df['COLE_COD_DANE_ESTABLECIMIENTO'].astype(str)
df['COLE_COD_DANE_SEDE'] = df['COLE_COD_DANE_SEDE'].astype(str)

# Función segura para arreglar acentos
def fix_text(x):
    if isinstance(x, str):
        try:
            return x.encode('latin1', errors='ignore').decode('utf-8', errors='ignore')
        except Exception:
            return x
    else:
        return x

# Aplicar solo a columnas de texto
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].apply(fix_text)

df.head()
```

```python
# Crear columnas nuevas a partir de PERIODO
df['AÑO'] = df['PERIODO'].astype(str).str[:4].astype(int)
df['SESION'] = df['PERIODO'].astype(str).str[4:].astype(int)

# Verificar
print(df[['PERIODO','AÑO','SESION']].drop_duplicates().sort_values('PERIODO'))
```

```python
# Definir valores válidos
valores_validos = ['A', 'B', 'OTRO', np.nan]

# Crear un filtro: filas donde COLE_CALENDARIO está en los válidos
mask_valid = df['COLE_CALENDARIO'].isin(valores_validos) | df['COLE_CALENDARIO'].isna()

# Aplicar filtro y eliminar las filas inválidas
df = df[mask_valid]

print("Número de filas después de limpieza:", len(df))
print("Valores únicos ahora:", df['COLE_CALENDARIO'].unique())
```

```python
import numpy as np

# Paso 1: Normalizar textos
df['COLE_CARACTER'] = df['COLE_CARACTER'].replace({
    'TCNICO/ACADMICO': 'TÉCNICO/ACADÉMICO',
    'ACADMICO': 'ACADÉMICO',
    'TCNICO': 'TÉCNICO',
    'NO APLICA': 'NO APLICA'
})

# Paso 2: Eliminar filas con valores numéricos atípicos
mask_valid = df['COLE_CARACTER'].isin(['TÉCNICO/ACADÉMICO','ACADÉMICO','TÉCNICO','NO APLICA']) | df['COLE_CARACTER'].isna()
df = df[mask_valid]

# Verificar resultados
print("Número de categorías:", df['COLE_CARACTER'].nunique())
print("Valores únicos:", df['COLE_CARACTER'].unique())
print("Número de filas después de limpieza:", len(df))
```

```python
# Lista de columnas que queremos limpiar
cols_dane = ['COLE_COD_DANE_ESTABLECIMIENTO','COLE_COD_DANE_SEDE']

for col in cols_dane:
    df[col] = (
        df[col]
        .astype(str)                # convertir a string
        .str.replace('.0','',regex=False)  # quitar el .0
        .replace('nan', np.nan)     # normalizar los NaN
    )
    # Convertir finalmente a número entero
    df[col] = df[col].astype(float).astype('Int64')

    # Verificar resultados
    print(f"Columna: {col}")
    print("Número de categorías:", df[col].nunique())
    print("Ejemplos de valores:", df[col].unique()[:10])
    print("-"*40)
```

```python
# Eliminar filas con el código 76
df = df[df['COLE_COD_DANE_ESTABLECIMIENTO'] != 76]

# Verificar que ya no exista
print("¿Todavía existe 76?:", (df['COLE_COD_DANE_ESTABLECIMIENTO'] == 76).any())
print("Número de filas después de limpieza:", len(df))
```

```python
# Limpiar columna 10 (COLE_COD_DEPTO_UBICACION)
df['COLE_COD_DEPTO_UBICACION'] = (
    df['COLE_COD_DEPTO_UBICACION']
    .astype(str)                # convertir a string
    .str.replace('.0','',regex=False)  # quitar el .0
    .replace('nan', np.nan)     # normalizar NaN si existiera
)

# Convertir finalmente a número entero
df['COLE_COD_DEPTO_UBICACION'] = df['COLE_COD_DEPTO_UBICACION'].astype(float).astype('Int64')

# Verificar resultados
print("Número de categorías:", df['COLE_COD_DEPTO_UBICACION'].nunique())
print("Ejemplos de valores:", df['COLE_COD_DEPTO_UBICACION'].unique()[:20])
```

```python
# Normalizar valores de la columna COLE_JORNADA
df['COLE_JORNADA'] = df['COLE_JORNADA'].replace({
    'MAANA': 'MAÑANA'
})

# Verificar resultados
print("Valores únicos después de limpieza:", df['COLE_JORNADA'].unique())
```

```python
# Eliminar filas con valores nulos en la columna COLE_CALENDARIO
df = df.dropna(subset=['COLE_CALENDARIO'])

# Verificar resultados
print("Número de valores nulos después de limpieza:", df['COLE_CALENDARIO'].isna().sum())
print("Valores únicos:", df['COLE_CALENDARIO'].unique())
```

```python
columna_21 = df.columns[20]

# Filtrar: mantener solo códigos válidos (numéricos de 2 dígitos o NaN)
df = df.loc[
    df[columna_21].apply(
        lambda x: (
            (isinstance(x, (int,float)) and not pd.isna(x) and len(str(int(x))) == 2)
            or pd.isna(x)
        )
    )
]

# Verificar resultados
print("Valores únicos después de limpieza:", df[columna_21].unique())
print("Número de filas después de limpieza:", len(df))
```

```python
# Detectar columnas numéricas y convertirlas a Int64
for col in df.select_dtypes(include=['float64','int64']).columns:
    df[col] = pd.to_numeric(df[col], errors="coerce").astype("Int64")

# Detectar columnas que parecen fechas (por nombre)
fecha_cols = [c for c in df.columns if "FECHA" in c.upper()]
for col in fecha_cols:
    df[col] = pd.to_datetime(df[col], errors="coerce")
```

```python
# Normalizar texto en todas las columnas tipo object
for col in df.select_dtypes(include=['object']).columns:
    df[col] = df[col].astype(str).str.strip().str.upper() \
                 .str.normalize("NFKD") \
                 .str.encode("ascii", errors="ignore") \
                 .str.decode("utf-8")
```

```python
# Exportar DataFrame limpio a CSV (más rápido y liviano)
ruta_salida_csv = '/content/drive/Shareddrives/GESTIÓN DE PROYECTOS PARA ANALÍTICA/Resultados_únicos_Saber_11_20260828_Limpio.csv'

df.to_csv(ruta_salida_csv, index=False, encoding='utf-8')

print("Archivo exportado en:", ruta_salida_csv)

```
